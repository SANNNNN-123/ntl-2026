Then **Add Input** this output on [`Qwen3.5-9B-base-Lora-twostage.ipynb`](Qwen3.5-9B-base-Lora-twostage.ipynb).

| --- | --- |
| transformers | 5.9.0 |
| bitsandbytes | 0.46.1 |
| accelerate | 1.6.0 |

If `5.9.0` is missing on PyPI for the image, change the pin to the newest 5.x that `pip download` accepts. Delete torch / nvidia / triton wheels; Kaggle already has those.


```python
!pip download "transformers==5.9.0" "bitsandbytes==0.46.1" "accelerate==1.6.0" "peft==0.15.2" \
    -d /kaggle/working/wheels
!rm -f /kaggle/working/wheels/torch-*.whl /kaggle/working/wheels/nvidia_*.whl /kaggle/working/wheels/triton-*.whl
!ls /kaggle/working/wheels
```

```python
from pathlib import Path
import json

wheels = Path("/kaggle/working/wheels")
files = sorted(p.name for p in wheels.glob("*.whl"))
need = ("transformers", "bitsandbytes", "accelerate", "peft")
missing = [n for n in need if not any(f.startswith(n) for f in files)]
assert not missing, f"download missed {missing}; files={files[:20]}"

manifest = {
    "purpose": "offline wheels for Qwen3.5-9B-base-Lora-twostage.ipynb",
    "packages": {
        "transformers": "5.9.0",
        "bitsandbytes": "0.46.1",
        "accelerate": "1.6.0",
        "peft": "0.15.2",
    },
    "wheels": files,
}
(wheels / "MANIFEST.json").write_text(json.dumps(manifest, indent=2))
print(f"{len(files)} wheels -> {wheels}")
print("\n".join(files))
```
